# Memory in RAG / GenAI Systems

### Simple Introduction

**Memory** means an AI system can remember and reuse useful information from earlier interactions.

Example:

```text
User: I prefer Python.
Later...
User: Which language should I use?
AI: Python would be a good choice for you.
```

## Why do we need memory?

Memory helps an AI system:

- remember previous conversation
- personalize answers
- continue tasks across multiple turns
- avoid asking the same information again
- reuse useful past information

Without memory, every request can feel like a new conversation.

## Main Types of Memory

### 1. Short-Term / Context Memory
Information currently visible inside the conversation.
EX:In the current chat, you say: “Make this reel energetic.” A few messages later, you say: “Also add fast cuts.” What are we talking about right now?

### 2. Conversational Memory
Stores previous chat messages.
EX:The assistant stores the recent editing conversation: “Use upbeat music → add captions → keep it under 30 seconds → use fast cuts.”     

Common forms:
- Full buffer → keep all messages
- Window buffer → keep only last few messages
EX:If the chat is long, it keeps only the latest few instructions, such as: “Use vertical format, add captions, and make the ending faster.”
- Summary memory → summarize older messages
EX:Instead of keeping 50 editing messages, it stores a summary: *“User wants a 30-second energetic travel reel with quick cuts, captions, and upbeat music.”*   to save input token cost

### 3. Long-Term Memory
Stores information across different sessions.
EX:Next week you upload another reel. The assistant still remembers that you usually prefer *short, fast-paced reels with captions*.         



## Episodic, Semantic, and Procedural Memory

### Episodic Memory
Remembers **events**.

Example:

It remembers a specific previous event: *“For the last Goa travel reel, the user selected version 2 with quick transitions and 25-second duration.”*                   
### Semantic Memory
Remembers **facts**.

Example:

It learns general facts about your preferences: *“User prefers cinematic reels, short captions, and upbeat music

### Procedural Memory
Remembers **how to perform a task**.

Example:

It remembers how you normally want reels edited: *1. Remove silent parts → 2. Select highlights → 3. Add transitions → 4. Add captions → 5. Add music → 6. Export 9:16.*

## Simple Comparison

| Memory Type | What it remembers | Example |
|---|---|---|
| Context | Current conversation | Previous message |
| Conversational | Chat history | Last 5 messages |
| Long-term | Information across sessions | User preference |
| Episodic | Events | Previous task |
| Semantic | Facts | User likes Python |
| Procedural | How to do something | Workflow steps |

## Memory-Augmented RAG

A GenAI system can use both RAG and memory.

```text
                 User Question
                      |
          -------------------------
          |                       |
          v                       v
      RAG Store              Memory Store
   Documents / Data        User / Chat History
          |                       |
          -----------   -----------
                    |
                    v
                  LLM
                    |
                    v
                  Answer
```

This gives answers that are both:

- grounded in external knowledge
- personalized using past information

````python
# Very simple memory example

memory = {
    "preferred_language": "Python",
    "explanation_style": "Simple"
}

question = "Which programming language should I use?"

print("Question:", question)
print("Retrieved memory:", memory["preferred_language"])
print("Answer: I would suggest", memory["preferred_language"])
````
